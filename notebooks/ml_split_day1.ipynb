```python
import pandas as pd

df = pd.DataFrame({
    "patient_id":["P001","P001","P002","P002","P003","P003","P004","P004"],
    "condition":["Normal","Normal","Adenoma","Adenoma","CRC","CRC","CRC","CRC"],
    "expression":[5,6,10,11,20,22,25,27]
})

df
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_df, test_df = train_test_split(
    df,
    test_size=0.25,
    random_state=42
)

print("Train:")
print(train_df)

print("\nTest:")
print(test_df)
```

```python
patients = df["patient_id"].unique()
print(patients)
```

```python
train_patients, test_patients = train_test_split(
    patients,
    test_size = 0.25,
    random_state = 3
)

print("Train patients", train_patients)
print("Test patients", test_patients)
```

```python
train_df = df[df["patient_id"].isin(train_patients)]
test_df = df[df["patient_id"].isin(test_patients)]

print("Train:")
print(train_df)

print("\nTest:")
print(test_df)
```
