```python
def check_age(age):
    if age >= 60:
        return "60 or above"
    else:
        return "under 60"
```

```python
def classify_condition(condition):
    if condition == "Normal":
        return("Normal tissue")
    elif condition == "Adenoma":
        return("Precancerous")
    elif condition == "CRC":
        return("Cancer")

result = classify_condition("Normal")
print(result)
    
```

```python
result = classify_condition("Adenoma")
print(result)
```

```python
result = classify_condition("CRC")
print(result)
```

```python
conditions = ["Normal", "Adenoma", "CRC"]

for condition in conditions:
    result = classify_condition(condition)
    print(result)
```
