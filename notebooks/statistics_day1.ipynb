```python
import statistics

normal = [5,6,5,7,6]
adenoma = [10,12,11,13,12]
crc = [20,25,22,30,28]

normal_mean = sum(normal) / len(normal)
normal_median = statistics.median(normal)
normal_sd = statistics.stdev(normal)

adenoma_mean = sum(adenoma) / len(adenoma)
adenoma_median = statistics.median(adenoma)
adenoma_sd = statistics.stdev(adenoma)

crc_mean = sum(crc) / len(crc)
crc_median = statistics.median(crc)
crc_sd = statistics.stdev(crc)

print(normal_mean, normal_median, normal_sd)
print(adenoma_mean, adenoma_median, adenoma_sd)
print(crc_mean, crc_median, crc_sd)
```
