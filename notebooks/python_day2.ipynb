```python
condition = "CRC"

if condition == "Normal":
    print("Normal tissue")
elif condition == "Adenoma":
    print("Precancerous lesion")
else:
    print("Cancer")
```

```python
conditions = ["Normal", "Adenoma", "CRC"]

for condition in conditions:
    if condition == "Normal":
        print("Normal tissue")
    elif condition == "Adenoma":
        print("Precancerous lesion")
    else:
        print("Cancer")
```
