```python
import pandas as pd

df = pd.read_csv("../data/metadata_practice.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df["condition"].value_counts()
```

```python
group_means = df.groupby("condition")["expression"].mean()
print(group_means)
```

```python
import statistics

group_std = df.groupby("condition")["expression"].std()
print(group_std)
```

```python
import matplotlib.pyplot as plt

group_means.plot(kind  = "bar")

plt.ylabel("Mean gene expression")
plt.title("Mean expression by group")

plt.savefig("../figures/Mean expression by group test.png")
plt.show()
```
