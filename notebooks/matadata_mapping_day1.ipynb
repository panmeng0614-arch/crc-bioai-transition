```python
excel_path = "../data/GSE201348/GSE201348_metadata_cleaned.xlsx"

table_s1 = pd.read_excel(excel_path,sheet_name="TableS1")
table_s2 = pd.read_excel(excel_path,sheet_name = "TableS2",header=1)
table_s3=pd.read_excel(excel_path,sheet_name="TableS3")
```

```python
table_s1 = table_s1.rename(columns={
    "Donor Study ID": "donor_id",
    "Disease Status": "disease_status"
})

table_s2 = table_s2.rename(columns={
    "Sample": "sample_id",
    "Donor": "donor_id",
    "GrossPathology": "gross_pathology",
    "PolypType": "polyp_type",
    "Dysplasia": "dysplasia"
})

table_s3 = table_s3.rename(columns={
    "General Paper ID": "sample_id",
    "GEO Title – snRNA": "geo_title_snrna"
})
```

```python
print(table_s1.columns)
print(table_s2.columns)
print(table_s3.columns)
```

```python
snrna_meta = table_s3[
   [
      "sample_id",
      "geo_title_snrna" 
   ] 
].copy()
```

```python
print(snrna_meta.shape)
```

```python
snrna_meta.head(10)
```

```python
snrna_meta["geo_title_snrna"].isna().sum()
```

```python
snrna_meta = snrna_meta[
    snrna_meta["geo_title_snrna"].notna()
].copy()
```

```python
print(snrna_meta.shape)
```

```python
snrna_meta["sample_id"].isna().sum()
```

```python
snrna_meta["sample_id"].duplicated().sum()
```

```python
snrna_meta[
    snrna_meta["sample_id"].duplicated(keep=False)
]
```

```python
snrna_meta = snrna_meta.merge(
    table_s2[
        [
            "sample_id",
            "donor_id",
            "gross_pathology",
            "polyp_type",
            "dysplasia"
         ]
    ],
    on="sample_id",
    how="left"
)
```

```python
snrna_meta["condition_raw"] = snrna_meta["gross_pathology"]
```

```python
condition_map = {
    "Unaffected" : "Normal",
    "Polyp" : "Polyp",
    "Adenocarcinoma" : "CRC"
}

snrna_meta["condition"] = snrna_meta["condition_raw"].map(condition_map)
```
