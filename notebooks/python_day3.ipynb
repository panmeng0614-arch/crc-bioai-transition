```python
patient1 = {
    "id": "P001",
    "age": 45,
    "condition": "Normal"
}

patient2 = {
    "id": "P002",
    "age": 57,
    "condition": "Adenoma"
}

patient3 = {
    "id": "P003",
    "age": 65,
    "condition": "CRC"
}

print(patient1)
```

```python
print(patient2)
```

```python
print(patient3)
```

```python
print(patient3["condition"] == "CRC")
```

```python
patients = [patient1, patient2, patient3]
print(patients)
```

```python
for patient in patients:
    print(patient)
```

```python
for patient in patients:
    print(patient["id"])
```

```python
for patient in patients:
    print(patient["condition"])
```

```python
for patient in patients:
    if patient["condition"] == "Normal": 
        print("Normal")
    elif patient["condition"] == "Adenoma":
        print("Precancerous")
    elif patient["condition"] == "CRC":
        print("Cancer")
    
    if patient["age"] <= 50:
        print("Age 50 or below")
    
```

```python
sample1 = {
    "sample_id": "N001",
    "patient_id": "P101",
    "condition": "Normal" 
}

sample2 = {
    "sample_id": "A001",
    "patient_id": "P102",
    "condition": "Adenoma"
}

sample3 = {
    "sample_id": "C001",
    "patient_id": "P103",
    "condition": "CRC"
}

sample4 = {
    "sample_id": "C002",
    "patient_id": "P104",
    "condition": "CRC"
}

samples = [sample1, sample2, sample3, sample4]

for sample in samples:
    if sample["condition"] == "CRC":
        print(sample["sample_id"], sample["patient_id"])
```
