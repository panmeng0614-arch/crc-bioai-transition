```python
import matplotlib.pyplot as plt
```

```python
conditions = ["Normal", "Adenoma", "CRC"]
mean_expression = [5.8, 11.6, 25.0]

plt.bar(conditions, mean_expression)

plt.xlabel("Condition")
plt.ylabel("Mean expression")
plt.title("Gene expression across disease stages")

plt.show()
```

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/metadata_practice.csv")

df.describe()
```

```python
group_means = df.groupby("condition")["expression"].mean()

print(group_means)
```

```python
group_means.plot(kind = "bar")
plt.ylabel("Mean expression")
plt.title("Mean expression by condition")

plt.savefig("../figures/expression_by_condition.png") 
plt.show()
```

```python
groups = ["Normal", "Adenoma", "CRC"]
sd_values = [0.8, 1.1, 4.1]

plt.bar(groups, sd_values)

plt.xlabel("group")
plt.ylabel("standard deviation")
plt.title("Standard deviation by group")

plt.show()
```
